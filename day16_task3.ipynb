```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

# create skewed population (not normal)
population = np.random.exponential(scale=2, size=10000)

# check population distribution
plt.hist(population, bins=50)
plt.title("Original Population (Highly Skewed)")
plt.show()
```

```python
sample_means = []

for i in range(1000):
    sample = np.random.choice(population, size=30)
    sample_means.append(np.mean(sample))
```

```python
plt.hist(sample_means, bins=40)
plt.title("Distribution of Sample Means (CLT Proof)")
plt.show()
```
