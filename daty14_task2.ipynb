```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
data = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 48, 30, 34, 42],
    'Salary': [25000, 32000, 120000, 150000, 110000, 170000, 115000, 40000, 50000, 90000]
})

data
```

```python
standard_scaler = StandardScaler()
data_standardized = pd.DataFrame(standard_scaler.fit_transform(data),columns=data.columns)
```

```python
minmax_scaler = MinMaxScaler()
data_normalized = pd.DataFrame(minmax_scaler.fit_transform(data),columns=data.columns)
data_standardized, data_normalized
```

```python
plt.figure()
plt.hist(data['Salary'])
plt.title("Salary Before Scaling")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python

plt.figure()
plt.hist(data_standardized['Salary'])
plt.title("Salary After Standardization")
plt.xlabel("Standardized Salary")
plt.ylabel("Frequency")
plt.show()
```
