```python
import pandas as pd
import numpy as np

data = {
    "Marks": [40, 45, 50, 55, 60, 65, 70, 1000]  
}

df = pd.DataFrame(data)
print(df)
```

```python
mean = df["Marks"].mean()
std = df["Marks"].std()

print("Mean:", mean)
print("Standard Deviation:", std)
```

```python
df["z_score"] = (df["Marks"] - mean) / std
print(df)
```

```python
outliers = df[np.abs(df["z_score"]) > 2]
print("Outliers:")
print(outliers)
```
