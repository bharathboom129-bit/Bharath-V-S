```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# normal data (heights)
heights = np.random.normal(loc=170, scale=10, size=1000)

df1 = pd.DataFrame({"Heights": heights})

print("Mean:", df1["Heights"].mean())
print("Median:", df1["Heights"].median())

sns.histplot(df1["Heights"], kde=True)
plt.title("Normal Distribution (Heights)")
plt.show()
```

```python
# right skewed data (income)
income = np.random.exponential(scale=50000, size=1000)

df2 = pd.DataFrame({"Income": income})

print("Mean:", df2["Income"].mean())
print("Median:", df2["Income"].median())

sns.histplot(df2["Income"], kde=True)
plt.title("Right-Skewed Distribution (Income)")
plt.show()
```

```python
# left skewed data (easy exam scores)
scores = 100 - np.random.exponential(scale=10, size=1000)

df3 = pd.DataFrame({"Scores": scores})

print("Mean:", df3["Scores"].mean())
print("Median:", df3["Scores"].median())
df2["Income_log"] = np.log(df2["Income"])
sns.histplot(df3["Scores"], kde=True)
plt.title("Left-Skewed Distribution (Exam Scores)")
plt.show()
```
